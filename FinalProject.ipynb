## 2. Background

Global access to electricity is a fundamental pillar of socioeconomic development and a core component of the United Nations Sustainable Development Goals (SDGs). Historically, electrification was viewed through a purely industrial lens, but contemporary research emphasizes its role as a "human right" essential for health, education, and poverty reduction. Despite its importance, universal access remains a long winded goal due to geographical barriers, political instability, and the high infrastructure costs associated with reaching remote rural populations.  

The Sustainable Development Goal 7 (SDG 7) initiative aims to ensure access to affordable, reliable, sustainable, and modern energy for all by 2030. To monitor progress toward this target, the World Bank and the Energy Sector Management Assistance Program (ESMAP) maintain a comprehensive global database on electrification rates.  

The SDG 7.1.1 Electrification Dataset provides standardized measures of the percentage of the population with access to electricity for over 200 countries and territories. This data is compiled from a variety of sources, including household surveys (such as the Demographic and Health Surveys) and census data. Global energy access observed from 1990 through 2023 shows a significant increase in global electrification, with the percentage of the population having access rising from roughly 71% in 1990 to over 91% by 2022, but also a persistent "rural-urban" gap, where urban areas maintain almost complete access while rural regions, continue to lag.

# Methodology and Modeling
To formally quantify the relationships, we will use statistical models to examine whether income groups are associated with electrification rates in 2013 and 2023. To examine if income group is associated with electricity access, we use ordinary least squares linear regression. Linear regression will model the relationship between a response variable and explanatory variables by estimating coefficients that minimize the sum of squared residuals between the observed and predicted values. In this model, income group is treated as a categorical variable in order to estimate the average difference in electrification rates between the World Bank income classifications.
### Model Structure

**Model 1: 2013 Electrification**

$$
Y_i = \beta_0 + \beta_1(\text{IncomeGroup}_i) + \varepsilon_i
$$

where:

- $Y_i$ is the electrification rate for country $i$ in 2013,
- $\beta_0$ is the intercept,
- $\beta_1$ represents the effect of a country's income group,
- $\varepsilon_i$ is the random error term.

**Model 2: 2023 Electrification**

$$
Y_i = \beta_0 + \beta_1(\text{IncomeGroup}_i) + \varepsilon_i
$$

where:

- $Y_i$ is the electrification rate for country $i$ in 2023,
- $\beta_0$ is the intercept,
- $\beta_1$ represents the effect of a country's income group,
- $\varepsilon_i$ is the random error term.

```python
import statsmodels.formula.api as smf

# model 1
model1 = smf.ols(formula='elec_2013 ~ C(IncomeGroup)', data=df_model).fit()

# model 2
model2 = smf.ols(formula='elec_2023 ~ C(IncomeGroup)', data=df_model).fit()
```

### Model Comparison

```python
comparing_models = pd.DataFrame({
    "Model": ["2013", "2023"],
    "R²": [model1.rsquared, model2.rsquared],
    "Adj. R²": [model1.rsquared_adj, model2.rsquared_adj],
    "AIC": [model1.aic, model2.aic],
    "BIC": [model1.bic, model2.bic]
})

comparing_models
```

To compare the income group and access to electricity relationship, we looked at the coefficient of determination R^2 for both models. The 2013 model had an R^2 value of 0.706, while the 2023 model had an R^2 value of 0.646. This shows that income group is still a strong predictor of electricity access, but it shows that the effect has decreased slightly over time, which may be a result of improvements in electrification among lower-income countries.

## K-Means
For this project, we are required to use a method beyond the scope of the course. When researching additional methods, I found K-Means, an unsupervised machine learning technique that identitifies natureal groupings in data sets. The K-Means class can be found in the Scikit-learn library, and will group similar observations together, allowing our project to explore whether countries naturally form clusters based on their electrification rates in 2013 and 2023.

```python
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler

#cluster variables
cluster_vars = df_model[['elec_2013', 'elec_2023']]

#standardizing the variables
cluster_scaled = StandardScaler().fit_transform(cluster_vars)

#KMeans
kmeans = KMeans(n_clusters=3, random_state=123, n_init=10)

df_model["Cluster"] = kmeans.fit_predict(cluster_scaled)

df_model.head()
```

### K-Means Scatterplot

```python
plt.figure(figsize=(8,6))

sns.scatterplot(data=df_model, x="elec_2013", y="elec_2023", hue="Cluster", palette="viridis")

plt.xlabel("Electrification Rate in 2013")
plt.ylabel("Electrification Rate in 2023")
plt.title("K-Means Clustering")
plt.legend(title="Cluster")

plt.show()
```

The K-Means clustering algorithm separated the countries into three main groups based on their electrification rates in 2013 and 2023. One group consists of countries that have almost complete access to electricity, another contains countries with moderate levels of access, and the last group is made up of countries with the lowest electrification rates. Overall, the graph shows that countries in all three clusters generally improved their access to electricity over the ten-year period. This supports the results from the regression models, which suggested that although lower-income countries still have lower electrification rates, many of them have made significant progress over time.
